## Carga de Datos y agrupación

```python
import pandas as pd

# Carga de datos
df =  pd.read_csv("/content/drive/MyDrive/Procesamiento del Lenguaje Natural/split1.mx-news.txt", sep = '\t')

# Impresión de primeras filas
print(df.head())

print("\nDistribución de etiquetas NER:")
print(df['Tag'].value_counts())
```

```python
def agrupar_oraciones(df):
    oraciones  = []
    grouped  =  df.groupby("Sentence #")
    for _, grupo in grouped:
        sentence = [(w, p, t) for w, p, t in zip(grupo["Word"], grupo["Pos"], grupo["Tag"])]
        oraciones.append(sentence)
    return oraciones

sentences  =  agrupar_oraciones(df)
print(f"Ejemplo de oración agrupada:\n {sentences[0]}")
```

## Extracción de características

```python
def word2features(sent, i):
    word, postag, _ = sent[i]

    # Caracteristica básicas definidas en el PPTX
    features = {
        'word.lower()': word.lower(),
        'postag': postag,
        'is_title': word.istitle(),
        'is_upper': word.isupper(),
        'is_digit': word.isdigit(),
    }

    if i > 0:
        word1, postag1, _ =  sent[i-1]
        features.update({
            '-1:word.lower()': word1.lower(),
            '-1:postag': postag1,
        })
    else:
        features['BOS'] =  True

    if i <  len(sent)-1:
        word1, postag1, _ =  sent[i+1]
        features.update({
            '+1:word.lower()': word1.lower(),
            '+1:postag': postag1,
        })
    else:
        features['EOS'] =  True

    return features
```
